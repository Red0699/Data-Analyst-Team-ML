```python
import requests

def logout(jwt):
    url = "https://arena.rone.dev/api/user/auth/logout"

    headers = {
        "Authorization": f"Bearer {jwt}"
    }

    try:
        respuesta = requests.post(
            url,
            headers=headers
        )

        datos = respuesta.json()

        if respuesta.status_code == 200 and datos.get("code") == 0:
            return datos

        return datos

    except Exception as e:
        return str(e)
```
